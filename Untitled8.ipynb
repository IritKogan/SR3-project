```python
# 1. Download the CelebA dataset via kagglehub
print("Downloading CelebA dataset... This might take a minute.")
path = kagglehub.dataset_download("jessicali9530/celeba-dataset")
print("Path to dataset files:", path)

# Locate the folder containing the actual images
image_dir = os.path.join(path, "img_align_celeba", "img_align_celeba")
if not os.path.exists(image_dir):
    image_dir = os.path.join(path, "img_align_celeba")
print(f"Targeting image directory: {image_dir}")

# 2. Define the Real CelebA Super-Resolution Dataset Class
class RealCelebASRDataset(Dataset):
    def __init__(self, image_dir, hr_size=64, lr_size=16, max_samples=10000):
        self.image_paths = sorted(glob.glob(os.path.join(image_dir, "*.jpg")))

        if len(self.image_paths) == 0:
            raise FileNotFoundError(f"No .jpg images found in {image_dir}. Check the directory path.")

        # Limit dataset size for faster training cycles on Colab
        if max_samples and max_samples < len(self.image_paths):
            self.image_paths = self.image_paths[:max_samples]

        print(f"Successfully indexed {len(self.image_paths)} real images for training.")

        # SR3 requires mapping pixel values from [0, 1] to [-1, 1]
        self.hr_transform = transforms.Compose([
            transforms.Resize((hr_size, hr_size)),
            transforms.ToTensor(),
            transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
        ])

        self.lr_transform = transforms.Compose([
            transforms.Resize((lr_size, lr_size)),
            transforms.ToTensor(),
            transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
        ])

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        img_path = self.image_paths[idx]
        img = Image.open(img_path).convert('RGB')

        hr_img = self.hr_transform(img)
        lr_img = self.lr_transform(img)

        return {"HR": hr_img, "LR": lr_img}

# 3. Instantiate Dataset and DataLoader
# Increase max_samples if you want to train on a larger portion of the dataset
dataset = RealCelebASRDataset(image_dir=image_dir, hr_size=64, lr_size=16, max_samples=10000)
data_loader = DataLoader(dataset, batch_size=16, shuffle=True, num_workers=2, drop_last=True)
```

```python
class PositionalEmbedding(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.dim = dim

    def forward(self, time):
        device = time.device
        half_dim = self.dim // 2
        embeddings = math.log(10000) / (half_dim - 1)
        embeddings = torch.exp(torch.arange(half_dim, device=device) * -embeddings)
        embeddings = time[:, None] * embeddings[None, :]
        embeddings = torch.cat((embeddings.sin(), embeddings.cos()), dim=-1)
        return embeddings

class Block(nn.Module):
    def __init__(self, in_ch, out_ch, time_emb_dim):
        super().__init__()
        self.time_mlp = nn.Linear(time_emb_dim, out_ch)
        self.conv1 = nn.Conv2d(in_ch, out_ch, 3, padding=1)
        self.conv2 = nn.Conv2d(out_ch, out_ch, 3, padding=1)
        self.relu = nn.ReLU()

    def forward(self, x, t_emb):
        h = self.relu(self.conv1(x))
        time_emb = self.relu(self.time_mlp(t_emb))
        h = h + time_emb[(..., ) + (None, ) * 2]
        h = self.relu(self.conv2(h))
        return h

class ConditionalUNet(nn.Module):
    def __init__(self, in_channels=6, out_channels=3, time_emb_dim=32):
        super().__init__()
        # in_channels=6 because we concatenate noisy_HR (3 ch) + upscaled_LR (3 ch)
        self.time_mlp = nn.Sequential(
            PositionalEmbedding(time_emb_dim),
            nn.Linear(time_emb_dim, time_emb_dim),
            nn.ReLU()
        )

        self.inc = nn.Conv2d(in_channels, 64, 3, padding=1)
        self.down1 = Block(64, 128, time_emb_dim)
        self.down2 = Block(128, 256, time_emb_dim)

        self.pool = nn.MaxPool2d(2)
        self.up1 = nn.ConvTranspose2d(256, 128, 2, stride=2)
        self.up2 = nn.ConvTranspose2d(128, 64, 2, stride=2)

        self.dec1 = Block(256, 128, time_emb_dim)
        self.dec2 = Block(128, 64, time_emb_dim)
        self.outc = nn.Conv2d(64, out_channels, 1)

    def forward(self, x, t):
        t_emb = self.time_mlp(t)

        x1 = self.inc(x)
        x2 = self.pool(self.down1(x1, t_emb))
        x3 = self.pool(self.down2(x2, t_emb))

        x = self.up1(x3)
        x = torch.cat([x, x2], dim=1)
        x = self.dec1(x, t_emb)

        x = self.up2(x)
        x = torch.cat([x, x1], dim=1)
        x = self.dec2(x, t_emb)

        return self.outc(x)
```

```python
class SR3Diffusion(nn.Module):
    def __init__(self, denoise_fn, timesteps=1000, loss_type='l1'):
        super().__init__()
        self.denoise_fn = denoise_fn
        self.timesteps = timesteps
        self.loss_type = loss_type

        beta = torch.linspace(1e-4, 0.02, timesteps, dtype=torch.float32)
        alpha = 1.0 - beta
        alpha_bar = torch.cumprod(alpha, dim=0)

        self.register_buffer('betas', beta)
        self.register_buffer('alphas_bar', alpha_bar)
        self.register_buffer('sqrt_alphas_bar', torch.sqrt(alpha_bar))
        self.register_buffer('sqrt_one_minus_alphas_bar', torch.sqrt(1.0 - alpha_bar))

    def q_sample(self, x_hr, t, noise=None):
        if noise is None:
            noise = torch.randn_like(x_hr)

        sqrt_alpha_bar_t = self.sqrt_alphas_bar[t].view(-1, 1, 1, 1)
        sqrt_one_minus_alpha_bar_t = self.sqrt_one_minus_alphas_bar[t].view(-1, 1, 1, 1)

        return sqrt_alpha_bar_t * x_hr + sqrt_one_minus_alpha_bar_t * noise

    def forward(self, x_hr, x_lr, t):
        noise = torch.randn_like(x_hr)
        x_noisy = self.q_sample(x_hr=x_hr, t=t, noise=noise)

        # Bicubic upscale low-res condition to match high-res dimensions
        x_lr_upscaled = F.interpolate(x_lr, size=x_hr.shape[-2:], mode='bicubic', align_corners=False)

        # Concatenate condition along channel dimension
        predicted_noise = self.denoise_fn(torch.cat([x_noisy, x_lr_upscaled], dim=1), t)

        if self.loss_type == 'l1':
            return F.l1_loss(noise, predicted_noise)
        else:
            return F.mse_loss(noise, predicted_noise)

    @torch.no_grad()
    def p_sample_loop(self, x_lr):
        batch_size = x_lr.shape[0]
        hr_shape = (batch_size, 3, x_lr.shape[2] * 4, x_lr.shape[3] * 4)

        img = torch.randn(hr_shape, device=device)
        x_lr_upscaled = F.interpolate(x_lr, size=hr_shape[-2:], mode='bicubic', align_corners=False)

        for i in tqdm(reversed(range(self.timesteps)), desc='Refinement Steps', total=self.timesteps):
            t = torch.full((batch_size,), i, device=device, dtype=torch.long)

            model_input = torch.cat([img, x_lr_upscaled], dim=1)
            pred_noise = self.denoise_fn(model_input, t)

            beta_t = self.betas[t].view(-1, 1, 1, 1)
            sqrt_alpha_bar_t = self.sqrt_alphas_bar[t].view(-1, 1, 1, 1)
            sqrt_one_minus_alpha_bar_t = self.sqrt_one_minus_alphas_bar[t].view(-1, 1, 1, 1)
            alpha_t = (1.0 - beta_t)

            mean = (1 / torch.sqrt(alpha_t)) * (img - (beta_t / sqrt_one_minus_alpha_bar_t) * pred_noise)

            if i > 0:
                noise = torch.randn_like(img)
                sigma_t = torch.sqrt(beta_t)
                img = mean + sigma_t * noise
            else:
                img = mean
        return img
```

```python
# Optimized Hyperparameters
EPOCHS = 50
TIMESTEPS = 1000
LR = 2e-4

# Initialize GradScaler for Mixed Precision (FP16)
scaler = torch.cuda.amp.GradScaler()

# Model Setup
unet = ConditionalUNet(in_channels=6, out_channels=3).to(device)
sr3 = SR3Diffusion(denoise_fn=unet, timesteps=TIMESTEPS, loss_type='l1').to(device)
optimizer = torch.optim.Adam(sr3.parameters(), lr=LR)

print("Starting OPMITIZED SR3 Training (with Mixed Precision)...")
for epoch in range(1, EPOCHS + 1):
    sr3.train()
    epoch_loss = 0
    progress_bar = tqdm(data_loader, desc=f"Epoch {epoch}/{EPOCHS}")

    for batch in progress_bar:
        optimizer.zero_grad(set_to_none=True) # set_to_none=True is slightly faster than zero_grad()

        x_hr = batch["HR"].to(device, non_blocking=True)
        x_lr = batch["LR"].to(device, non_blocking=True)

        t = torch.randint(0, TIMESTEPS, (x_hr.shape[0],), device=device).long()

        # Runs the forward pass with Autocasting (Mixed Precision)
        with torch.cuda.amp.autocast():
            loss = sr3(x_hr, x_lr, t)

        # Backward pass with scaled gradients
        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

        epoch_loss += loss.item()
        progress_bar.set_postfix({"Loss": loss.item()})

    print(f"Epoch {epoch} finished. Avg Loss: {epoch_loss / len(data_loader):.4f}")
```

### Save the Trained Model

After training, it's crucial to save the model's state dictionary. This allows you to reload the trained model later without having to retrain it, which is especially useful for long training processes.

```python
# Define a path to save the model
model_save_path = "sr3_model.pth"

# Save the state dictionary of the trained model
torch.save(sr3.state_dict(), model_save_path)

print(f"Trained model saved to: {model_save_path}")
```

```python
# Evaluate model
sr3.eval()

# Grab real validation samples
sample_batch = next(iter(data_loader))
x_lr_sample = sample_batch["LR"][:2].to(device)
x_hr_sample = sample_batch["HR"][:2].to(device)

print("Generating 4x Super-Resolution faces via Repeated Refinement...")
generated_hr = sr3.p_sample_loop(x_lr_sample)

# Helper function to denormalize images back to [0, 1] range for Matplotlib
def denorm(tensor):
    return torch.clamp((tensor + 1.0) / 2.0, 0.0, 1.0)

# Generate a bicubic interpolation upscale to serve as your comparison baseline
bicubic_baseline = F.interpolate(x_lr_sample, size=generated_hr.shape[-2:], mode='bicubic')

# Plot the results
fig, axes = plt.subplots(2, 3, figsize=(12, 8))
labels = ["Low Res Input (16x16)", "Bicubic Interpolation", "SR3 Result (64x64)"]

for i in range(2):
    # Low-Res
    axes[i, 0].imshow(denorm(x_lr_sample[i]).cpu().permute(1, 2, 0))
    axes[i, 0].set_title(labels[0])
    axes[i, 0].axis('off')

    # Bicubic
    axes[i, 1].imshow(denorm(bicubic_baseline[i]).cpu().permute(1, 2, 0))
    axes[i, 1].set_title(labels[1])
    axes[i, 1].axis('off')

    # SR3 Generated output
    axes[i, 2].imshow(denorm(generated_hr[i]).cpu().permute(1, 2, 0))
    axes[i, 2].set_title(labels[2])
    axes[i, 2].axis('off')

plt.tight_layout()
plt.show()
```
